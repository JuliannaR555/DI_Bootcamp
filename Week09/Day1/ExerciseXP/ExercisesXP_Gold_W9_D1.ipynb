**Prompt Task**

**Original Prompt**
> A shop sells pencils at $0.75 each. If Alice buys 6 pencils and pays with a $5 bill, how much change does she get? Let’s solve this step-by-step.  
> 6 pencils × $0.75 = $4.75  
> $5.00 - $4.75 = $0.50

**Instructions**

2) Rewrite the prompt with the correct Chain-of-Thought.

## Exercise 2: Choose the Right Prompt Pattern

Chosen pattern: Few-Shot Classification with label definitions + rubric
Why: Short customer messages are ambiguous; few curated examples + boundary rules improve consistency, generalization, and tie-break behavior vs. zero-shot.

### Title: Few-Shot Classification Prompt with Label Definitions
You are a message classifier. Choose ONE label from:
- Billing Issue: charges, refunds, invoices, payment methods.
- Technical Support: bugs, errors, broken features, app not working.
- Account Access: login, password, 2FA, locked account.
- Other: anything else.

Decision Rules:
- If message mentions money/charges/refund → Billing.
- If it reports malfunction or error codes → Technical.
- If it mentions login/password/2FA/locked → Account Access.
- If none apply → Other.
Return JSON {"label": "<one_of_above>", "rationale": "<1 short sentence>"}.

Examples:
Q: "I was billed twice this month."
A: {"label":"Billing Issue","rationale":"Double charge is a billing problem."}

Q: "The app crashes when I open settings."
A: {"label":"Technical Support","rationale":"Crash indicates a technical problem."}

Q: "I forgot my password and can’t log in."
A: {"label":"Account Access","rationale":"Password/login is account access."}

Q: "Can I change the color theme?"
A: {"label":"Other","rationale":"Preference request, not billing/tech/access."}

The pattern aligns the model via explicit label semantics, decision rules, and consistent JSON output, reducing drift and improving reproducibility across varied phrasings.

## Exercise 3: Use AlignedCoT to Compare Reasoning Paths

### Title: AlignedCoT Prompt — Two Paths + Consistency Check
You will reason via TWO distinct paths, then compare and select the consistent answer.

Problem:
Small pot = $2; Medium = $4; Large = $6.
She buys 2 small, 3 medium, 1 large. What is the total cost?

Path A (Unit-Price Decomposition):
- Compute each subtotal = quantity × unit price.
- Sum the subtotals.

Path B (Weighted Sum Framing):
- Vector of quantities [2,3,1]; prices [2,4,6].
- Compute dot product: (2×2) + (3×4) + (1×6).

Comparison Step:
- If Path A and Path B match, output that total.
- If mismatch, re-check arithmetic and explain the correction.

Output format:
- Path A total = $X
- Path B total = $Y
- Final answer = $Z (reason)

Now solve.

Expected final:

- Path A: $4 + $12 + $6 = $22
- Path B: 2 × 2 + 3 × 4 + 1 × 6 = 4 + 12 + 6 = $22

Final answer: $22

### Interpretation:
Two differently framed paths reduce slip errors; the agreement step acts as a built-in verification.

## Exercise 4: Design a Multi-Step Document Pipeline

### Title: Stage 1 — Domain Identification

Role: Academic triage assistant.

Input: Title + abstract (or full text snippet).

Task: Classify the paper’s PRIMARY domain from:

["biology","physics","computer science","mathematics","medicine","social science","other"].

Output JSON: {"domain":"<label>","confidence":0–1,"evidence":"<short phrase>"}.

If confidence < 0.70, set "domain":"other" and add evidence explaining ambiguity.

Text:
<<<PAPER_METADATA_AND_ABSTRACT>>>

### Title: Stage 2 — Contribution Extraction

Role: Research summarizer.

Input: Abstract + domain from Stage 1.

Task: Extract up to 3 key contributions (verbs + objects), each ≤ 20 words.

Output JSON: {"domain":"<from_stage1>","contributions":["...","...","..."]}.

If no clear contributions found, return {"contributions":["Not stated in abstract."]}.

Text: <<<ABSTRACT>>>

### Title: Stage 3 — Follow-up Research Question

Role: Research ideation assistant.

Input: Domain + contributions from Stage 2.

Task: Generate 1 precise follow-up research question aligned to the domain; include measurable outcome or dataset type.

Output JSON: {"question":"<one sentence>","rationale":"<≤18 words>"}.

Constraints: Avoid speculation beyond provided abstract; tie the question to at least one extracted contribution.

Where to use conditional logic:

- If Stage 1 confidence < 0.70, route to human review before Stage 3.
- If Stage 2 returns “Not stated…”, Stage 3 should ask for full paper or methods section instead of proposing a question.

Context chaining: Pass domain and contributions as immutable inputs to Stage 3 to prevent drift.

## Exercise 5: Role Prompting to Reduce Bias

### Title: Basic Prompt (Likely to Bias)

Given a user’s skills and interests, suggest 3 career paths.

User profile: <<<PROFILE_TEXT>>>

Return a short list with 1-line reasons.

### Title: Revised Role-Prompt to Reduce Bias

Role: Career counselor trained in fair, inclusive guidance.

Instructions:
- Base suggestions ONLY on stated skills/interests. Ignore demographic attributes (gender, age, ethnicity, disability, etc.).
- Provide 4 diverse career paths across different industries.
- For each path, explain the skill match and a next-step (cert/course/portfolio).
- Add a “Fairness Check” bullet stating how stereotypes were avoided.

Input:
<<<PROFILE_TEXT>>>

Output:
- Bulleted list with 4 options; each: role, 1-line rationale, 1 concrete next step.
- Final bullet: “Fairness Check: …”

### Why this improves fairness:
The role primes impartiality; explicit exclusion of protected attributes, diversity requirement, and Fairness Check make biases visible and correctable.
