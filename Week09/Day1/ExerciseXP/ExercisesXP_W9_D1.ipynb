## Exercise 1: Rewrite and Optimize a Vague Prompt

Issues with the original prompt:

1. No target audience — doesn't specify who the LinkedIn post is for (e.g., knowledge workers, students, freelancers).
2. No brand voice or tone — no guidance on FlowNest's voice (friendly, focused, science-backed).
3. No platform/format constraints — LinkedIn favors scannable structure, hook, hashtags, and CTA; none are specified.

Interpretation:
By adding role, specific instructions, and clear format constraints, you reduce ambiguity and guide the model toward a LinkedIn-optimized, on-brand post.

### Optimized Prompt

[ROLE] Act as a B2B social media copywriter for a productivity startup called FlowNest.

[GOAL] Create a LinkedIn post that promotes FlowNest’s new focus app to busy knowledge workers.

[INSTRUCTIONS]
- Start with a 1-sentence hook that names a common focus problem.
- Give 3 actionable, evidence-informed tips that FlowNest enables (e.g., focus sessions, gentle nudges, progress analytics).
- Close with a clear CTA to try the app.

[FORMAT/TONE/LENGTH]
- Use short paragraphs + bullet points.
- Tone: friendly, pragmatic, and science-aware.
- Include 2–3 relevant hashtags (e.g., #DeepWork, #Productivity).
- Max 120–150 words.

[REUSABILITY]
Use this exact format whenever we promote productivity tools on LinkedIn for FlowNest.

## Exercise 2: Multi-Part Prompt for Quiz Generation

### Complete Multi-Part Prompt

[ROLE] You are an instructional designer for 7th-grade science (ages 11–13).

[INPUT] Article text about volcanic eruptions:
<<<ARTICLE_TEXT>>>

[TASKS]
1) Summarize the article in exactly 2 simple bullet points (plain language, short sentences).
2) Generate 3 multiple-choice questions (MCQs). Each MCQ must include:
   - 1 correct answer and 2 plausible distractors.
   - A brief, kid-friendly explanation for the correct answer.
3) Maintain a friendly tone and simple vocabulary suitable for 11–13-year-olds.

[OUTPUT FOR GOOGLE SLIDES]
- Slide 1: Title = "Volcanic Eruptions — Quick Summary"; 2 bullets with the summary.
- Slides 2–4: One MCQ per slide.
  - Title = "Question X"
  - List options A, B, C on separate lines.
  - After the options, add a line: "Answer & Why:" with the correct option letter + short explanation.
- Slide 5: "Teacher Notes" — list the 3 learning objectives covered in simple terms.

[CONSTRAINTS]
- No extra commentary outside slide content.
- Keep each bullet under 14 words.
- Avoid jargon; if needed, define it in 6 words or fewer.

Comparison to the vague prompt:

- Defined structure/output (Slides with titles, bullets, MCQ pattern) prevents free-form, messy responses.
- Age appropriate tone and vocabulary avoids overly technical or too-advanced language.
- Explicit constraints (exact counts, word limits, distractor rules) reduce irrelevant or low-quality content.

## Exercise 4: Match Prompt to Purpose

### Complete Prompt (Structured Style)

[ROLE] You are a customer support assistant for an e-commerce company.

[OBJECTIVE] Resolve customer issues efficiently and politely.

[TURN STRUCTURE]
1) Greeting + intent confirmation (1 sentence).
2) Information collection checklist (ONLY ask what’s missing): order ID, email, item name, issue type (delay/damage/return/other), preferred resolution (refund/replacement/status).
3) Validation step: repeat back key details in one sentence; ask “Is this correct?”
4) Action step (choose one): track order, initiate return/replacement, escalate to human.
5) Closure: confirm next steps and expected timeline.

[POLICIES]
- Be concise and empathetic.
- Never request sensitive data beyond email and order ID.
- If policy conflict: explain briefly and offer alternatives.

[OUTPUT FORMAT]
- Use short paragraphs and bullet lists.
- Bold the chosen action label (e.g., **Action:** Initiate replacement).

Why this matches the style:

1. Fixed sequence of steps and a checklist enforce a predictable format.
2. Output formatting rules (bullets, bold labels) standardize responses.

Why Structured over others:

- Compared to Conversational, it reduces rambling and speeds resolution.
- Compared to Exploratory, it avoids ideation and drives actions.
- Compared to Functional only, it still collects context before executing.

## Exercise 6: Hallucination Spotting and Mitigation

### 1) Precise “no hallucination” prompt:

[ROLE] You are summarizing a peer-reviewed article on climate change and marine biodiversity.

[INSTRUCTIONS]
- Use ONLY the information present in the provided article text.
- Do NOT infer numbers, projections, or claims not explicitly stated.
- If a detail is absent, write: “Not stated in the article.”

[OUTPUT]
- 3 bullet points summarizing key findings with citations (section/page).

### 2) Verification-focused rewrite:

Include ONLY claims explicitly stated in the article. 

If uncertain or not found, write: “Information not available in the article.”

For any quantitative claim, cite the exact sentence location (section/page).

Return 3 bullets max; no external sources or estimates.

3) Mitigation strategies used:

- Source-bound instructions (“use ONLY the provided text”) + fallback phrasing for missing info.
- Citation requirement (section/page) to force traceability and reduce unverifiable claims.

4) High-risk domains:

- Healthcare — hallucinated treatments or dosages can cause patient harm and legal liability.
- Legal — fabricated statutes/case law can mislead decisions, create compliance risks, and erode trust.

## Prompting Exercises — Interpretation & Conclusion

### Big Picture
Well-structured prompts (Role, Audience, Purpose, Input, Directives, Structure) turn vague asks into **reliable, reusable outputs**.

### Key Takeaways
1) Vague → Specific: role + audience + channel format = on-brand copy.
2) Education: slide schema + exact counts (2 bullets, 3 MCQs) = classroom-ready.
3) Context for numbers: role/audience/purpose + sections elicit figures & timeframes.
4) Match style to purpose: Structured steps beat rambling in support workflows.
5) Control output: exact counts, word caps, must-mention fields raise compliance.
6) Reduce hallucinations: source-bound rules + citations + “Not stated” fallback.

### RAPIDS Checklist
Role • Audience • Purpose • Input (allowed/forbidden) • Directives (steps/acceptance criteria) • Structure/Style (tone, format, limits)

### Quality Gate
Structure clear? Counts/limits explicit? Data + timeframes stated? Traceable (citations/fallbacks)?

### One-Sentence Takeaway
Great prompts make intent computable.
